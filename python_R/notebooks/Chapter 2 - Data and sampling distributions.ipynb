# Sampling Distribution of a Statistic

### Python

```python
loans_income = pd.read_csv(LOANS_INCOME_CSV).squeeze('columns')

sample_data = pd.DataFrame({
    'income': loans_income.sample(1000),
    'type': 'Data',
})

sample_mean_05 = pd.DataFrame({
    'income': [loans_income.sample(5).mean() for _ in range(1000)],
    'type': 'Mean of 5',
})

sample_mean_20 = pd.DataFrame({
    'income': [loans_income.sample(20).mean() for _ in range(1000)],
    'type': 'Mean of 20',
})

results = pd.concat([sample_data, sample_mean_05, sample_mean_20])
print(results.head())
```

```python
g = sns.FacetGrid(results, col='type', col_wrap=1, 
                  height=2, aspect=2)
g.map(plt.hist, 'income', range=[0, 200000], bins=40)
g.set_axis_labels('Income', 'Count')
g.set_titles('{col_name}')

plt.tight_layout()
plt.show()
```

### R
R replicates the same simulation logic using `sample`, `tapply`, and `ggplot2` faceting to compare the raw distribution and sampling distributions of means.

```python
%%R
# take a simple random sample
samp_data <- data.frame(income=sample(loans_income, 1000), 
                        type='data_dist')

# take a sample of means of 5 values
samp_mean_05 <- data.frame(
  income = tapply(sample(loans_income, 1000*5), 
                  rep(1:1000, rep(5, 1000)), FUN=mean),
  type = 'mean_of_5')

# take a sample of means of 20 values
samp_mean_20 <- data.frame(
  income = tapply(sample(loans_income, 1000*20), 
                  rep(1:1000, rep(20, 1000)), FUN=mean),
  type = 'mean_of_20')

# bind the data.frames and convert type to a factor
income <- rbind(samp_data, samp_mean_05, samp_mean_20)
income$type <- factor(income$type, 
                     levels=c('data_dist', 'mean_of_5', 'mean_of_20'),
                     labels=c('Data', 'Mean of 5', 'Mean of 20'))
```

```python
%%R
ggplot(income, aes(x=income)) +
  geom_histogram(bins=40) +
  facet_grid(type ~ .)
```

### Summary: Sampling Distribution of a Statistic

| Operation | Python | R | Notes |
|---|---|---|---|
| Random sampling from data | `Series.sample()` | `sample()` | Same statistical intent: repeated random draws from the observed data. |
| Build sampling means | list comprehension + `.mean()` | `tapply(..., FUN=mean)` | Python is iteration-centric; R uses grouped vectorized aggregation. |
| Compare distributions | `sns.FacetGrid()` + `plt.hist` | `ggplot2::facet_grid()` + `geom_histogram()` | Both produce aligned small multiples for comparison. |

Python tends to express simulation as explicit data-frame assembly, while R often favors vectorized transformations with concise statistical idioms.

# The Bootstrap

### Python

```python
results = []
for nrepeat in range(1000):
    sample = resample(loans_income)
    results.append(sample.median())
results = pd.Series(results)
print('Bootstrap Statistics:')
print(f'original: {loans_income.median()}')
print(f'bias: {results.mean() - loans_income.median()}')
print(f'std. error: {results.std()}')
```

### R
As the calculation uses random samples, results will vary between runs.

```python
%%R
stat_fun <- function(x, idx) median(x[idx])
boot_obj <- boot(loans_income, R=1000, statistic=stat_fun)

boot_obj
```

### Summary: The Bootstrap

| Operation | Python | R | Notes |
|---|---|---|---|
| Resampling engine | `sklearn.utils.resample()` | `boot::boot()` | Python performs manual loop orchestration; R wraps the full bootstrap workflow. |
| Target statistic | `sample.median()` | custom `stat_fun` + `median()` | Both compute median over bootstrap resamples. |
| Reported outputs | manual `bias` and `std. error` calculation | `boot` object summary | R provides standardized output; Python offers explicit control. |

Python is transparent and flexible because each step is explicit, while R is more compact through a specialized bootstrap interface.

# Confidence Intervals

### Python

```python
print(loans_income.mean())
np.random.seed(seed=3)  
# create a sample of 20 loan income data
sample20 = resample(loans_income, n_samples=20, replace=False)
print(sample20.mean())
results = []
for nrepeat in range(500):
    sample = resample(sample20)
    results.append(sample.mean())
results = pd.Series(results)

confidence_interval = list(results.quantile([0.05, 0.95]))
ax = results.plot.hist(bins=30, figsize=(4, 3))
ax.plot(confidence_interval, [55, 55], color='black')
for x in confidence_interval:
    ax.plot([x, x], [0, 65], color='black')
    ax.text(x, 70, f'{x:.0f}', 
            horizontalalignment='center', verticalalignment='center')
ax.text(sum(confidence_interval) / 2, 60, '90% interval',
        horizontalalignment='center', verticalalignment='center')

meanIncome = results.mean()
ax.plot([meanIncome, meanIncome], [0, 50], color='black', linestyle='--')
ax.text(meanIncome, 10, f'Mean: {meanIncome:.0f}',
        bbox=dict(facecolor='white', edgecolor='white', alpha=0.5),
        horizontalalignment='center', verticalalignment='center')
ax.set_ylim(0, 80)
ax.set_ylabel('Counts')

plt.tight_layout()
plt.show()
```

```python
np.random.seed(seed=3)
# create a sample of 20 loan income data
sample20 = resample(loans_income, n_samples=20, replace=False)

results = []
for nrepeat in range(500):
    sample = resample(sample20)
    results.append(sample.mean())
results = pd.Series(results)

confidence_interval = list(results.quantile([0.05, 0.95]))
ax = results.plot.hist(bins=30, figsize=(4, 3), color='C1')
ax.plot(confidence_interval, [55, 55], color='black', linestyle='--')
for x in confidence_interval:
    ax.plot([x, x], [0, 60], color='black')
ax.text(82000, 50, 
        f'90% CI\n[{confidence_interval[0]:.0f}, {confidence_interval[1]:.0f}]',
       fontsize='small')

confidence_interval = list(results.quantile([0.025, 0.975]))
ax = results.plot.hist(bins=30, figsize=(4, 3))
ax.plot(confidence_interval, [65, 65], color='black', linestyle='--')
for x in confidence_interval:
    ax.plot([x, x], [0, 70], color='black')
ax.text(82000, 65, 
        f'95% CI\n[{confidence_interval[0]:.0f}, {confidence_interval[1]:.0f}]',
       fontsize='small')
# ax.text(sum(confidence_interval) / 2, 264, '95 % interval',
#         horizontalalignment='center', verticalalignment='center')

meanIncome = results.mean()
ax.plot([meanIncome, meanIncome], [0, 50], color='black', linestyle='--')
ax.text(meanIncome, 5, f'Mean: {meanIncome:.0f}',
        bbox=dict(facecolor='white', edgecolor='white', alpha=0.5),
        horizontalalignment='center', verticalalignment='center')
ax.set_ylim(0, 80)
ax.set_xlim(37000, 102000)
ax.set_xticks([40000, 50000, 60000, 70000, 80000])
ax.set_ylabel('Counts')

# plt.tight_layout()
# plt.show()
```

### R
R bootstraps the sample mean and extracts a basic confidence interval with `boot.ci`, then overlays interval markers on a histogram.

```python
%%R
# R version for figure 2-9 not available
set.seed(7)
sample20 <- sample(loans_income, 20)
sampleMean <- mean(sample20)

stat_fun <- function(x, idx) mean(x[idx])
boot_obj <- boot(sample20, R=500, statistic=stat_fun)
boot_ci <- boot.ci(boot_obj, conf=0.9, type='basic')
X <- data.frame(mean=boot_obj$t)
ci90 <- boot_ci$basic[4:5]
ci <- data.frame(ci=ci90, y=c(9, 11))
ci
ggplot(X, aes(x=mean)) +
    geom_histogram(bins=40, fill='#AAAAAA') +
    geom_vline(xintercept=sampleMean, linetype=2) +
    geom_path(aes(x=ci, y=10), data=ci, linewidth=2) +
    geom_path(aes(x=ci90[1], y=y), data=ci, linewidth=2) +
    geom_path(aes(x=ci90[2], y=y), data=ci, linewidth=2) +
    annotate('text', x=sampleMean, y=20, label='Sample mean', size=6) +
    annotate('text', x=sampleMean, y=8, label='90% interval', size=6) +
    theme_bw() + 
    labs(x='', y='Counts')
```

### Summary: Confidence Intervals

| Operation | Python | R | Notes |
|---|---|---|---|
| Draw base sample | `resample(..., n_samples=20, replace=False)` | `sample(loans_income, 20)` | Both start from a finite sample before bootstrap inference. |
| Bootstrap means | repeated `resample(sample20)` + `.mean()` | `boot(sample20, statistic=mean)` | Python computes explicitly; R delegates to `boot`. |
| CI extraction | `quantile([0.05, 0.95])` / `quantile([0.025, 0.975])` | `boot.ci(..., type='basic')` | Different interval constructions but same inferential goal. |
| Interval annotation | matplotlib line/text layers | `ggplot2` geoms + `annotate()` | Both annotate uncertainty directly on distribution plots. |

Python makes interval definitions explicit via quantiles, while R encourages formal bootstrap-interval APIs that encode methodology choices.

# Normal Distribution
## Standard Normal and QQ-Plots

### Python
The package _scipy_ has the function (`scipy.stats.probplot`) to create QQ-plots. The argument `dist` specifies the distribution, which is set by default to the normal distribution.

```python
fig, ax = plt.subplots(figsize=(4, 4))

norm_sample = stats.norm.rvs(size=100)
stats.probplot(norm_sample, plot=ax)

plt.tight_layout()
plt.show()
```

### R
R creates the same QQ diagnostic with `qqnorm` and adds a reference line using `abline`.

```python
%%R
norm_samp <- rnorm(100)
qqnorm(norm_samp, main='', xlab='Quantile of normal distribution', ylab='z-score')
abline(a=0, b=1, col='grey')
```

### Summary: Standard Normal and QQ-Plots

| Operation | Python | R | Notes |
|---|---|---|---|
| Generate normal sample | `stats.norm.rvs(size=100)` | `rnorm(100)` | Equivalent normal sampling in both ecosystems. |
| QQ-plot construction | `stats.probplot(..., plot=ax)` | `qqnorm()` | Both compare sample quantiles to theoretical normal quantiles. |
| Reference line | built into `probplot` output | `abline(a=0, b=1)` | R adds the reference line explicitly. |

Python centralizes QQ diagnostics in a single SciPy call, while R composes the plot in two clear base-graphics steps.

# Long-Tailed Distributions

### Python

```python
sp500_px = pd.read_csv(SP500_DATA_CSV)

nflx = sp500_px.NFLX
nflx = np.diff(np.log(nflx[nflx>0]))

fig, ax = plt.subplots(figsize=(4, 4))
stats.probplot(nflx, plot=ax)

plt.tight_layout()
plt.show()
```

### R
R applies the same log-return transformation and evaluates tail behavior with `qqnorm` plus a reference line.

```python
%%R
nflx <- sp500_px[,'NFLX']
nflx <- diff(log(nflx[nflx>0]))
qqnorm(nflx)
abline(a=0, b=1, col='grey')
```

### Summary: Long-Tailed Distributions

| Operation | Python | R | Notes |
|---|---|---|---|
| Extract series | `sp500_px.NFLX` | `sp500_px[,'NFLX']` | Same data selection with language-specific indexing style. |
| Log-return transform | `np.diff(np.log(...))` | `diff(log(...))` | Nearly identical mathematical transformation. |
| Tail check with QQ-plot | `stats.probplot()` | `qqnorm()` + `abline()` | Both reveal deviations from normality in the tails. |

The statistical workflow is almost identical here; the practical difference is mainly API style and plotting defaults.

## Exponential Distribution

### Python

```python
sample = stats.expon.rvs(scale=5, size=100)

pd.Series(sample).plot.hist()
plt.show()
```

### R
R draws exponential random values with `rexp`, using `rate` (inverse of scale) rather than Python's `scale` parameterization.

```python
%%R
rexp(n=100, rate=.2)
```

### Summary: Exponential Distribution

| Operation | Python | R | Notes |
|---|---|---|---|
| Draw exponential sample | `stats.expon.rvs(scale=5, size=100)` | `rexp(n=100, rate=.2)` | Same model with inverse parameterization (`scale = 1/rate`). |

The core difference is parameter naming (`scale` vs `rate`), a common practical source of confusion when moving between Python and R.
